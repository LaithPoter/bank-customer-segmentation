## EDA Summary & Business Insights

**Key observations from the analysis:**

- The strongest correlations in the data are between Monthly_Income and Account_Balance, Age and Work_Experience, and Age and Account_Balance — older customers tend to have more work experience and higher balances.
- Retired and older customers hold the highest account balances, likely due to accumulated pension income and a more settled lifestyle, while traders show the highest monthly income and generally high balances as well.
- The majority of customers are married (~67%) and mostly in their 30s — the data is somewhat imbalanced on this point.
- Baghdad and Basra account for the largest share of the bank's customers compared to other governorates.
- Most customers have less than 5 years of work experience and are not yet graduates, which partly explains why students cluster around lower balances.
- As account balance increases, the share of married customers and females within that balance range also increases.

**Business Insight:**
The data suggests clear segments worth targeting with different campaigns: (1) retirees with stable income and high balances — savings and investment products, (2) traders with high income — premium banking services and business loans, (3) young/student customers with low balances — incentive products or entry-level credit cards, (4) large families with limited income — installment loans or family savings plans. This early segmentation motivates applying K-Means to formally identify these groups rather than relying only on visual observations.

#DATA Preprocessing

```python
from sklearn.preprocessing import StandardScaler , OneHotEncoder
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df_model= df.drop("Customer_ID" , axis = 1)
#doesn't matter and could effect negatively
```

```python
df_model['Gender'] = df_model['Gender'].map({'Male':0, 'Female':1})
df_model['Ever_Married'] = df_model['Ever_Married'].map({'No':0, 'Yes':1})
df_model['Graduated'] = df_model['Graduated'].map({'No':0, 'Yes':1})
#manual encoding
```

```python
df_model = pd.get_dummies(df_model, columns=['Profession','City'], drop_first=True)
#one-hot encoding for multi-category with dummies to make them 0 or 1 , and drop first to delete others and keeps one.
```

```python
df_model.info()
#now there's no any categories columns.
```

```python
scaler = StandardScaler()
scaled_array = scaler.fit_transform(df_model)
#scaling all data to make all the numbers of the same range
```

```python
df_scaled = pd.DataFrame(scaled_array, columns=df_model.columns)#get the real names back to the data.
df_scaled.head()
#after scaling we made it back as pandas file to get the right names to the scaled files, and show the first 5 of them.
```

```python
groups = []
silhouette = []
K_range = range(2, 11)
for i in K_range:
  km = KMeans(n_clusters=i,init='k-means++',random_state=42)
  km.fit(df_scaled)
  labels = km.predict(df_scaled)
  groups.append(km.inertia_)
  silhouette.append(silhouette_score(df_scaled, labels))
#we train here to see the right clustring trial.
```

```python
plt.figure(figsize=(10,5))
plt.plot(K_range,groups,color='red')
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('groups')
plt.show()
#plot the elbow method.
```

```python
plt.figure(figsize=(6,4))
plt.plot(K_range, silhouette, marker='o', color='green')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score')
plt.show()
#plot the silhouette equation
#reveals that at 7 num of clusters is the better solution for this data.
```

```python
best_k = 7 #based on the silhouete and elbow method.
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(df_scaled)
df['Cluster'].value_counts()
#the best sol that we trained on.
```

```python
plt.figure(figsize=(10, 6))
x = df_scaled.values
y_kmeans = df['Cluster'].values

income_idx = df_scaled.columns.get_loc('Monthly_Income')
balance_idx = df_scaled.columns.get_loc('Account_Balance')

plt.scatter(x[y_kmeans == 0, income_idx], x[y_kmeans == 0, balance_idx], s=100, c='red', label='Cluster 1')
plt.scatter(x[y_kmeans == 1, income_idx], x[y_kmeans == 1, balance_idx], s=100, c='blue', label='Cluster 2')
plt.scatter(x[y_kmeans == 2, income_idx], x[y_kmeans == 2, balance_idx], s=100, c='green', label='Cluster 3')
plt.scatter(x[y_kmeans == 3, income_idx], x[y_kmeans == 3, balance_idx], s=100, c='cyan', label='Cluster 4')
plt.scatter(x[y_kmeans == 4, income_idx], x[y_kmeans == 4, balance_idx], s=100, c='magenta', label='Cluster 5')
plt.scatter(x[y_kmeans == 5, income_idx], x[y_kmeans == 5, balance_idx], s=100, c='orange', label='Cluster 6')
plt.scatter(x[y_kmeans == 6, income_idx], x[y_kmeans == 6, balance_idx], s=100, c='purple', label='Cluster 7')
plt.scatter(kmeans.cluster_centers_[:, income_idx], kmeans.cluster_centers_[:, balance_idx], s=300, c='yellow', label='Centroids')

plt.title('Clusters of Bank Customers')
plt.xlabel('Monthly Income (scaled)')
plt.ylabel('Account Balance (scaled)')
plt.legend()
plt.show()
#now here we could see the clusters plot and group 6 are the most number of groups have samples.
```
